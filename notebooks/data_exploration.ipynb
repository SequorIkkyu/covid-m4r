```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/uk_final.csv")
df["date"] = pd.to_datetime(df["date"])

df.head()
```

```python
df.info()
df.describe()
```

```python
print(df["date"].min(), df["date"].max())
print(len(df))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))
plt.plot(df["date"], df["new_cases"])
plt.title("Raw daily cases")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(df["date"], df["new_cases"], alpha=0.3, label="daily")
plt.plot(df["date"], df["cases_7d"], label="7-day avg")

plt.legend()
plt.grid()
plt.title("UK COVID cases")

plt.show()
```

```python
df_model = df[(df["date"] >= "2021-11-01") & (df["date"] <= "2022-03-31")]
```

```python
plt.figure(figsize=(10,4))
plt.plot(df_model["date"], df_model["cases_7d"])
plt.title("Selected modelling window (Omicron wave)")
plt.grid()
plt.show()
```
