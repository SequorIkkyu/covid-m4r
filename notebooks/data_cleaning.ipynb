```python
import pandas as pd

# =========================
# DAILY DATA
# =========================
df_daily = pd.read_csv("../data/ukhsa-chart-download.csv")

df_daily = df_daily[["date", "metric_value"]].copy()
df_daily.columns = ["date", "new_cases"]

df_daily["date"] = pd.to_datetime(df_daily["date"])
df_daily = df_daily.sort_values("date")

df_daily["new_cases"] = df_daily["new_cases"].clip(lower=0)
df_daily["new_cases"] = df_daily["new_cases"].fillna(0)

df_daily = df_daily.reset_index(drop=True)

# =========================
# 7-DAY SMOOTHED DATA
# =========================
df_7d = pd.read_csv("../data/ukhsa-chart-download-7day.csv")

df_7d = df_7d[["date", "metric_value"]].copy()
df_7d.columns = ["date", "cases_7d"]

df_7d["date"] = pd.to_datetime(df_7d["date"])
df_7d = df_7d.sort_values("date")

df_7d["cases_7d"] = df_7d["cases_7d"].clip(lower=0)
df_7d["cases_7d"] = df_7d["cases_7d"].fillna(0)

df_7d = df_7d.reset_index(drop=True)

# =========================
# SAVE
# =========================
df_daily.to_csv("../data/uk_clean_daily.csv", index=False)
df_7d.to_csv("../data/uk_clean_7day.csv", index=False)

print("Cleaned files saved!")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(df_daily["date"], df_daily["new_cases"], alpha=0.3, label="daily")
plt.plot(df_7d["date"], df_7d["cases_7d"], label="7-day avg")

plt.legend()
plt.grid()
plt.title("Daily vs 7-day smoothed cases")

plt.show()
```
